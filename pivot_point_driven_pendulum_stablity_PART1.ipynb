2. **Define the pendulum:**

    a. *Define terms:*

        l = length of rigid pendulum
        g = 9.81 m/s**2 = gravity
        m = mass of the pendulum
        mg = Force on the pendulum in the y-plane
        ω_0 = ... angular velocity
        ω_d = ... angular velocity
        v = velocity
        t = time
        θ = angle measured from the downward vertical (=0 when straight down)

    b. *Equations:*

  $$x_p(t) = -l \sin(\omega_d t)$$
  $$y_p(t) = l (\cos(\omega_d t) - 1)$$

*For these equations, I can take the derivitives of them to get the pivot points velocity and accleration, which will affect how the bob moves.I can then use these to get the position of the bob, with respect to some vertical angle using phi*.

First Derivitive
  $$dx_p(t) = -l w_d cos(w_d t)$$
  $$dy_p(t) = -l w_d sin(w_d t)$$
Second Derivitive
  $$ddx_p(t) = l w_d^2 sing(w_d t)$$
  $$ddy_p(t) = -l w_d^2 cos(w_d t)$$

*Bob's Position:*
  $$x(t) = x_p(t) + l sin(ϕ(t))$$
  $$y(t) = x_p(t) - l cos(ϕ(t))$$
First Derivitive
  $$dx(t) = dx_p(t) + l cos(ϕ(t)) * dϕ(t)$$
  $$dy(t) = dy_p(t) + l sin(ϕ(t)) * dϕ(t)$$
Second Derivitive
  $$ddx(t) = ddx_p(t) ...$$
  $$ddy(t) = ddy_p(t) ...$$

*Define Energies:*
  $$T = 1/2 * m * (dx(t) + dy(t))^2$$
  $$V = mgy(t)$$
*Define the Lagrangian:*
  $$L = T - V$$
*DO the Euler-Lagrange Equation: to get the exact equation of motion*
