# 2. Define parameters, initial conditions, and equations

### **Parameters:**
`ω_d` for the driven angular velcity,
`gamma` for damping,
`g` for gravitational accleration,
`l` for the length of the pendulum,
`m` for the mass of the bob,
`t` for time, and
`ω_0 = np.sqrt(g/l)` for the natural frequency

Bob parameters include `θ(t)`, `̇θ(t)`,`̈θ(t)` for the bob's angular displacement, velocity, and acceleration, respectively. Natural frequency `ω_0 = np.sqrt(g/l)`.

*Time Parameter:*

`t_min = 0` for initial time,
`t_max = 100` for final time, and
`t_span = (t_min, t_max)` for the time span.

---
### **Initial Conditions:**

`theta_0 = 0 + np.pi - np.arcsin(2*gamma / omega_d)` for the initial angle of the pendulum bob,
`ang_vel_0 = omega_d` for the initial angular velocity, and
`y_0 = [theta_0, omega_d]`for the bob's initial position and "VELOCITY...", and
`x_p0` and `y_p0` for the pivot point's initial position.

---
### **Equations:**
*Pivot Point position:*
$$x_p(t) = -l * sin(\omega_d * t)$$
$$y_p(t) = l * (cos(\omega_d * t) - 1)$$

*Bob position:*

$$x(t) = x_p(t) + l * sin(\theta(t))$$
$$y(t) = x_p(t) - l * cos(\theta(t))$$


````python
# parameters
omega_d = 2.0
gamma = 0.1
g = 9.81
l = 1.0
m = 1.0
t = sp.symbols('t')
omega_0 = np.sqrt(g/l)

# time parameter
t_min = 0
t_max = 100
t_span = (t_min, t_max)

# initial conditions
theta_0 = 0  #+ np.pi - np.arcsin(2*gamma / omega_d)
ang_vel_0 = omega_d
y_0 = [theta_0, ang_vel_0]

x_p0 = 0
y_p0 = 0

# equations for positions
# pivot point position
x_p = x_p0 - l * sp.sin(omega_d * t)
y_p = y_p0 + l * (sp.cos(omega_d * t) - 1)
# bob position
x = x_p + l * sp.sin(theta_0)
y = y_p - l * sp.cos(theta_0)

#test print
print(x_p)
print(y_p)
print(x)
print(y)
````

````python
# Defining the Lagrangian
T = 0.5 * m * (x_vel + y_vel)**2
V = m * g * y_eq
L = T - V
# test printing
#print(f"T: {T}")
#print(f"V: {V}")
#print(f"L: {L}")

# take the partial theta of L with respect to theta and dtheta
L_dtheta = get_derivitive(L, theta)
L_ddottheta = get_derivitive(L_dtheta, theta)
# take the time derivitive of L_ddottheta
L_ddottheta_t = get_derivitive(L_ddottheta, t)
# test printing
print(f"L_dtheta: {L_dtheta}")
print(f"L_ddottheta: {L_ddottheta}")
print(f"L_ddottheta_t: {L_ddottheta_t}")
````
